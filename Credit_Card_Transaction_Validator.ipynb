```python
def validate_transaction(balance,amount,daily_limit,used_today):
  if amount <=0:
    print("Access declined: Amount must be greater than 0")
    return
  if amount > balance:
    print("Transcation declined: Amount is less than balance")
    return

  remaining_daily_limit = daily_limit - used_today
  if amount > remaining_daily_limit:
    print("Transcation declined: Daily transcation limit exceeded ")
    return

  if balance - amount < 1000
    print("Transaction Declined: Minimum balance of Rs. 1,000 must remain")

  new_balance= balance - amount
  new_daily_limit = remaining_daily_limit - amount

  print("Transcation Approved!!")
  print(f"Remaining Balnce: {new_balance:.0f}")
  print(f"Remaining daily limit: {new_daily_limit:.0f}")

balance = float(input("Enter account balance: "))
amount = float(input("Enter transaction amount: "))
daily_limit = float(input("Enter daily transaction limit: "))
used_today  = float(input("Enter today's used amount: "))

validate_transaction(balance, amount, daily_limit, used_today)
```
